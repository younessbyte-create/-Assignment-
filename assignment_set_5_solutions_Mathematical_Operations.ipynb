### 2. Fit a polynomial curve of degree 4 on the drug sales data and plot the curve on original dataset

``` 
url = 'https://raw.githubusercontent.com/selva86/datasets/master/a10.csv'
ts = np.genfromtxt(url, delimiter=',', dtype='float', usecols=[1])
ts = ts[1:]
```

````python
url = 'https://raw.githubusercontent.com/selva86/datasets/master/a10.csv'
ts = np.genfromtxt(url, delimiter=',', dtype='float', usecols=[1])
ts = ts[1:]
````

````python
# Solution
a = np.arange(len(ts))
poly = np.polyfit(a, ts, 4)

# coefs
print("Polynomial fit \n", poly)
````

````python
# Fitted values to smoothen the data
cubic_fit = np.polyval(poly, a)
cubic_fit[:5]
````

````python
# Plot the curve on original dataset
import matplotlib.pyplot as plt
%matplotlib inline

plt.figure(figsize=(10,7))
plt.plot(ts)
plt.plot(cubic_fit)
plt.title("Apple Close Price", size=20)
plt.show()
````
